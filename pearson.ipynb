```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

# 矩阵数据
data = np.array([
    [59, 15, 7, 4, 2],
    [7, 109, 7, 2, 2],
    [4, 7, 11, 2, 2],
    [4, 4, 2, 1, 0],
    [2, 7, 7, 0, 4]
])

# 标签
categories = ['ENCOURAGE', 'PRESSURE', 'CRITICAL', 'MISLEADING', 'HEURISTIC']
row_totals = [87, 127, 26, 11, 10]
col_totals = [76, 142, 34, 9, 10]

# 计算准确率
correct = sum(data[i][i] for i in range(5))
total = 261
accuracy = correct / total

# 温和配色（YlGnBu 风格 + 自定义柔和渐变）
# 方案1: 使用内置温和配色 'YlOrRd' 的柔化版
colors = [(0.95, 0.95, 0.98), (0.85, 0.92, 0.95), (0.6, 0.8, 0.85), (0.4, 0.6, 0.7), (0.2, 0.4, 0.5)]
# 或者使用更现代的柔和蓝绿配色
cmap_gentle = LinearSegmentedColormap.from_list('Gentle', colors, N=256)

# 更简单的方案：使用内置温和配色 'viridis' 的柔和版
# 或者使用 'Blues' 但降低饱和度
# 这里使用 'YlGnBu'（黄绿蓝）温和渐变
cmap_gentle = plt.cm.YlGnBu  # 温和的黄绿蓝配色

# 创建图形
fig, ax = plt.subplots(figsize=(12, 9))

# 绘制热力图（温和配色）
im = ax.imshow(data, cmap=cmap_gentle, aspect='auto', vmin=0, vmax=110)

# 设置刻度
ax.set_xticks(np.arange(5))
ax.set_yticks(np.arange(5))
ax.set_xticklabels(categories, rotation=45, ha='right', fontsize=11, fontweight='medium')
ax.set_yticklabels(categories, fontsize=11, fontweight='medium')

# 在每个格子中显示数值
for i in range(5):
    for j in range(5):
        value = data[i, j]
        # 根据背景色自动选择文字颜色（值较大时背景深，用白色字）
        text_color = 'white' if value > 45 else '#2c3e50'
        text = ax.text(j, i, str(value),
                       ha="center", va="center",
                       color=text_color,
                       fontsize=12, fontweight='bold')

# 添加行和（右侧）
for i in range(5):
    ax.text(5.15, i, f'{row_totals[i]}', ha='left', va='center',
            fontsize=11, fontweight='bold', color='#8b4513', style='italic')

# 添加列和（底部）
for j in range(5):
    ax.text(j, 5.1, f'{col_totals[j]}', ha='center', va='top',
            fontsize=11, fontweight='bold', color='#8b4513', style='italic')

# 添加准确率副标题
ax.set_title(f'Strategy Selection Accuracy\n {correct}/{total} = {accuracy:.2%}',
             fontsize=16, fontweight='bold', pad=25, color='#2c3e50')

# 添加边框
for spine in ax.spines.values():
    spine.set_visible(True)
    spine.set_linewidth(1.2)
    spine.set_color('#7f8c8d')

# 添加颜色条
cbar = plt.colorbar(im, ax=ax, shrink=0.8, aspect=30)
cbar.set_label('Number of Samples', fontsize=11, fontweight='medium', color='#2c3e50')
cbar.ax.tick_params(labelsize=10, colors='#2c3e50')

# 设置坐标轴标签
ax.set_xlabel('Prediction', fontsize=13, fontweight='bold', color='#2c3e50', labelpad=15)
ax.set_ylabel('Ground Truth', fontsize=13, fontweight='bold', color='#2c3e50', labelpad=15)

# 添加行和/列和注释
ax.text(5.5, 5.3, 'Row Sum →', ha='right', va='bottom', 
        fontsize=9, color='#8b4513', style='italic')
ax.text(5.5, 5.55, '↓ Column Sum', ha='right', va='bottom', 
        fontsize=9, color='#8b4513', style='italic')

# 调整布局
plt.tight_layout()
plt.savefig('strategy_selection_confusion_matrix.png', dpi=200, bbox_inches='tight', facecolor='white')
plt.show()

print(f"✅ 矩阵已保存为 strategy_selection_confusion_matrix.png")
print(f"📊 准确率: {correct}/{total} = {accuracy:.4f} ({accuracy:.2%})")
print(f"\n配色方案: YlGnBu（黄绿蓝温和渐变）")
print(f"📌 坐标轴: Ground Truth (行) vs Prediction (列)")
```
